```python
model = SentenceTransformer("all-MiniLM-L6-V2")
```

```python
embeddings = model.encode(sql_data)
```

```python
embeddings = np.array(embeddings).astype('float32')
```

```python
print("Embedding Shape:", embeddings.shape)
```

```python
vector_db = {
    "texts": sql_data,
    "embeddings": embeddings
}

print("Vector database created")
print("Total records:", len(vector_db["texts"]))
```

```python
# Step 6: Cosine similarity function
def cosine_similarity(query_vector, stored_vectors):
    
    # Dot product
    dot_product = np.dot(stored_vectors, query_vector)
    
    # Magnitudes
    query_norm = np.linalg.norm(query_vector)
    stored_norms = np.linalg.norm(stored_vectors, axis=1)
    
    # Cosine similarity
    similarity = dot_product / (stored_norms * query_norm)
    
    return similarity
```

```python
def search_vector_db(query, top_k = 3):
    query_embedding = model.encode(query)
    query_embedding = np.array(query_embedding).astype("float32")

    scores = cosine_similarity(query_embedding, vector_db["embeddings"])

    top_indices = np.argsort(scores)[::-1][:top_k]
    
    print(f"\nQuery: {query}\n")

    for i, idx in enumerate(top_indices):
        print(f"Match {i+1} | Score: {scores[idx]:.4f}")
        print(vector_db["texts"][idx])
        print("-"*50)
```

```python
search_vector_db("What is the difference between inner join and right join?")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
model_name = "google/flan-t5-large"

tokenizer = AutoTokenizer.from_pretrained(model_name)
llm = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("LLM loaded successfully")
```

```python
#creating a prompt template
def build_prompt(query, retrieved_chunks):
    
    context = "\n".join(retrieved_chunks)

    prompt = f"""
You are an expert SQL interview assistant.

Answer the question clearly using the context provided, below there will be 3 context so learning them answer the question.

Context:
{context}

Question:
{query}

Give a clear SQL interview answer in 2-3 sentences.
Explain the concept properly.
"""
    
    return prompt
```

```python
#Genearating answer from LLM model.
def generate_answer(prompt):
    
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True)

    outputs = llm.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.3
    )

    answer = tokenizer.decode(outputs[0], skip_special_tokens=True)

    return answer
```

```python
#connecting whole retrieval and LLM together.
def rag_sql_assistant(query, top_k=3):

    # Embed query
    query_embedding = model.encode(query)
    query_embedding = np.array(query_embedding).astype("float32")

    # Calculate similarity
    scores = cosine_similarity(query_embedding, vector_db["embeddings"])

    # Get top results
    top_indices = np.argsort(scores)[::-1][:top_k]

    retrieved_chunks = [vector_db["texts"][i] for i in top_indices]

    # Build prompt
    prompt = build_prompt(query, retrieved_chunks)

    # Generate answer
    answer = generate_answer(prompt)

    print("\nRetrieved Context:\n")
    for chunk in retrieved_chunks:
        print("-", chunk)

    print("\nFinal Answer:\n")
    print(answer)
```

```python
rag_sql_assistant("What is right join")
```
