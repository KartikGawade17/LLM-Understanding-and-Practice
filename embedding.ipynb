```python
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

A = np.array([1, 2])
B = np.array([2, 4])
C = np.array([1, 0])

print("A vs B:", cosine_similarity(A, B))
print("A vs C:", cosine_similarity(A, C))
```

```python
import faiss
```

```python
dimension = 4
np.random.seed(42)
data = np.random.random((5, dimension)).astype('float32')
```

```python
index = faiss.IndexFlatL2(dimension)

# Add vectors to index
index.add(data)

print("Number of vectors in index:", index.ntotal)
```

```python
# Create query vector
query = np.random.random((1, dimension)).astype('float32')

# Search for 2 nearest neighbors
k = 2
distances, indices = index.search(query, k)

print("Distances:", distances)
print("Indices:", indices)
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
import faiss
```

```python
# Load small but powerful embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Sample sentences
sentences = [
    "I love artificial intelligence",
    "Machine learning is amazing",
    "I enjoy studying AI",
    "The weather is very hot today",
    "It is sunny outside"
]

# Generate embeddings
embeddings = model.encode(sentences)

print("Embedding shape:", embeddings.shape)
```

```python
# Convert to float32 for FAISS
embeddings = embeddings.astype("float32")

# Normalize vectors (important for cosine similarity)
faiss.normalize_L2(embeddings)
```

```python
dimension = embeddings.shape[1]

# Use inner product index
index = faiss.IndexFlatIP(dimension)

# Add embeddings
index.add(embeddings)

print("Total vectors in index:", index.ntotal)
```

```python
query = "what is machine learning"

query_embedding = model.encode([query]).astype("float32")
faiss.normalize_L2(query_embedding)

k = 2
distances, indices = index.search(query_embedding, k)

print("Query:", query)
print("\nTop matches:\n")

for idx in indices[0]:
    print(sentences[idx])
```
